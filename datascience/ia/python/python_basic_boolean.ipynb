Asigna el valor de True a una variable y muestralo.

```python
administrador = "rosanela"
usuario = input("Usuario: ")
es_admin = f"{usuario}: Ingresando al sistema......" if usuario == administrador else f"{usuario}: Usuario no registrado"
print(es_admin)
```

Asigna un valor False a una variable y muéstralo.

```python
administrador = "rosanela"
password = "frutillita"
acceso_concedido = False


ingresando_rol = input("Usuario: ")
ingresando_password = input("Contraseña: ")

if ingresando_rol == administrador and ingresando_password == password:
    acceso_concedido = True
    print("Ingresando......")
else:    
    print(f"Acceso permitido: {acceso_concedido}")
    print("Error: No estás registrado")
```

 Convierte un número a booleano (0 = False, cualquier otro = True).

```python
valores = [1, 23, 0, 5, 454, 0, 0, -1]


for valor in valores:
    print(f"Falso {valor}" if valor == 0 else f"Verdad {valor}")

for valor in valores:
    resultado = bool(valor)
    print(f"{valor} convertido es: {resultado}")
```

 Convierte una cadena vacía y una no vacía a booleano

```python
cadenas = ["", "Frutilla", " "]
for cadena in cadenas:
    booleano = bool(cadena)
    print(f"{cadena} --> {booleano}")
```

Verifica si una lista está vacía usando booleano

```python
cadenas = [[],["Javascript"],"frutilla", " ", "", "python", 2]
for elemento in cadenas:
    if isinstance(elemento,(list, str)):
        esta_vacio = not bool(elemento)
        print(f"Esta vacio '{elemento}'?: {esta_vacio}")
    else:
        print(f"El elemento {elemento} es de tipo {type(elemento).__name__} y no aplica  'vacio'")
    
```
