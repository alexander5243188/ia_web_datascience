Enteros (int), Calcula la diferencia entre dos enteros ingresados por el usuario.

```python
try:
    numero_1 = int(input("Ingresa el primer número entero: "))
    numero_2 = int(input("Ingresa el segundo número entero: "))
    diferencia = numero_1 - numero_2
    print(f" {numero_1} - {numero_2} = {diferencia}")
except ValueError:
    print("Error: Ingresar un valor valido")
```

Enteros (int), Multiplica tres enteros y muestra el resultado.

```python
try:
    numero_1 = int(input("Ingrese el primer valor entero: "))
    numero_2 = int(input("Ingrese el segundo valor entero: "))
    numero_3 = int(input("Ingrese el tercer valor entero: "))
    multiplica = numero_1 * numero_2 * numero_3
    print(f"{numero_1} * {numero_2} * {numero_3} = {multiplica}")
except ValueError as e:
    print(f"ERROR: {e}")
```

Enteros (int), Calcula la división entera y el residuo de dos números.

```python
try:
    dividendo = float(input("Ingrese el dividendo: (el númeroa dividir) "))
    divisor = float(input("Ingrese divisor: (el numero por el que dividira) "))
    if divisor == 0:
        print(f"No se puede dividir por {divisor}")
    else:
        division_entera = int(dividendo) // int(divisor)
        residuo = dividendo % divisor
        prueba = (division_entera * divisor )+residuo
        print(f"{dividendo} / {divisor} = {division_entera}")
        print(f"{dividendo} % {divisor} = {residuo}")
        print(f"{prueba} --> {dividendo}")
except ValueError as e:
    print(f"ERROR: {e}")
```

Enteros (int), Eleva un número entero a la potencia de otro.

```python
try:
    base = int(input("Ingresa un entero para la base: "))
    exponente = int(input("Ingresa un entero para el exponente: "))
    elevado_potencia = base * exponente
    print(f"{base} ** {exponente} = {elevado_potencia}")
except ValueError as e:
    print(f"ERROR {e}")
```

Flotantes (float), convierte un número entero a flotante y multiplicalo

```python
try:
    numero_entero = int(input("Ingresa el numero entero: "))
    numero_float = float(input("Ingresa el numero flotante:"))

    convertido = float(numero_entero)
    multiplicacion = convertido * numero_float

    print(f"{numero_entero} --> {convertido} ")
    print(f"{convertido} * {numero_float} = ", convertido*numero_float)
    
except ValueError as e:
    print(f"Error {e}")
```

Flotantes (float), Redonde un número flotante a 2 decimales

```python
try:
    numero = float(input("Ingrese un valor: "))
    redondeado = round(numero, 2)
    print(f"{numero} --> {redondeado}")
except ValueError as e:
    print(f"Error: {e}")
```

Flotantes (float), Calcula la división de dos enteros y muestra el resultado como float

```python
try:
    dividendo = int(input("Primer valor: "))
    divisor = int(input("Segundo valor: "))
    division_entera = dividendo // divisor
    temp = division_entera
    resultado_float = float(temp)
    print(f"{dividendo} / {divisor} = {division_entera} --> {resultado_float}")

except ValueError as e:
    print(f"Error  {e}")
except ZeroDivisionError:
    print("Error, no se puede dividir por cero")
```

Flotantes (float),  Calcula el valor absoluto de un numero flotante negativo

```python
try:
    valor = float(input("Ingrese un valor: "))
    if valor < 0:
        print(f"{valor} -->Valor Absoluto --> {abs(valor)}")
    else:
        print("Ingresa un valor valido")
except ValueError as e:
    print(f"ERORR: {e}")
```

Complejos (complex), crea un numero complejo y muestra su parte real y su parte imaginaria

```python
try:
    numero_complejo = complex(input("Ingresa un número complejo (ej. 3+4j, -5, 2.5j): "))
    parte_real = numero_complejo.real
    parte_imaginaria = numero_complejo.imag
    print(f"\nNúmero ingresado: {numero_complejo}")
    print(f"--> Parte real: {parte_real}")
    print(f"--> Parte imaginaria: {parte_imaginaria}")
    
except ValueError as e:
    print(f"Error: {e}")
```

Complejos (complex), suma de dos numeros complejos

```python
try:
    primer_numero = complex(input("Ingresa el primer número complejo Ej: 4+5j"))
    segundo_numero = complex(input("Ingresa el segundo complejo Ej: 4+5j"))
    print(f"{primer_numero} + {segundo_numero} = ", primer_numero + segundo_numero)
except ValueError  as e:
    print(f"Error: {e}")
```

Complejos (complex), calcula el conjugado de un numero complejo

```python
try:
    numero = complex(input("Ingresa un numero complejo: Ej: 3+5j"))
    print(f"{numero} conjugado --> {numero.conjugate()}")
except ValueError as e:
    print(f"ERROR: {e}")
```
