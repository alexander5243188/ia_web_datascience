Mi primera ventana

```python
import tkinter as tk

# 1. Crear la raiz (ventana principal)
root = tk.Tk()
root.title("Mi Primera GUI")
root.geometry("300x200") # ancho y alto

# 2. crea un widget (Etiqueta)
label = tk.Label(root, text="¡Hola, experto en Python!", fg="blue")

# 3. Empaquetar el widget (Hacerlo visible)
label.pack(pady=20)

# 4. Iniciar el bucle de escucha de eventos
root.mainloop()
```

Nivel Básico: "El Contador"
Crea una ventana que tenga una etiqueta mostrando el número 0. Debajo, coloca un botón que diga "Incrementar". Cada vez que el usuario presione el botón, el número en la etiqueta debe aumentar en 1.

```python
import tkinter as tk

def incremento():
    global contador
    contador += 1
    etiqueta.config(text=f"{contador}")


contador = 0

app = tk.Tk()
app.title="Contador basico"
app.geometry("200x150")

etiqueta = tk.Label(app, text=contador, font=("Arial", 24))
etiqueta.pack(pady=10)

boton = tk.Button(app, text="Incrementar", command=incremento)
boton.pack()

app.mainloop()
```

Nivel Intermedio: "Cambiador de Colores", Crea una ventana con un campo de entrada (Entry) y un botón. El usuario debe escribir el nombre de un color en inglés (ej. "red", "blue", "green"). Al presionar el botón, el color de fondo de la ventana principal debe cambiar al color escrito. Pista: Usa root.config(bg=color).

```python
import tkinter as tk

def cambiar():
    color = entrada.get().lower().strip()
    try:
        app.config(bg=color)
        mensaje.config(text=f"Color actual: {color}", fg="black")
    except tk.TclError:
        mensaje.config(text="¡Color no valido!", bg="red")

app = tk.Tk()
app.title("Cambiador de color")
app.geometry("300x200")

mensaje = tk.Label(app, text="Escribe un color en ingles (Ej; 'red', 'blue'): ")
mensaje.pack(pady=10)

entrada = tk.Entry(app)
entrada.pack(pady=5)

# Vincular la tecla 'Enter' para cambiar el color también
entrada.bind('<Return>', lambda event: cambiar())

boton = tk.Button(app, text="Cambiar fondo", command=cambiar)
boton.pack(pady=10)

app.mainloop()
```
