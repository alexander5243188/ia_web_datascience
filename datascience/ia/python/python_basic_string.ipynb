
Cadena (str), extrae una subcadena usando slicing

```python
cadena = str(input("Ingresa una cadena: "))
longitud = len(cadena)

if not cadena.strip(): # si la cadena esta vaciao tiene espacios
    print("Error: No se admite entradas vacias.")
else:
    print(f"La cadena tiene {longitud} caracteres, Indices de 0 a {longitud -1}")    
    try:
        corte_inicio = int(input("Indice de inicio:"))
        corte_final = int(input("Indice de final: "))

        resultado = cadena[corte_inicio:corte_final]

        print("-"*20)
        print(f"Cadena original {cadena}")
        print(f"Carte [{corte_inicio}:{corte_final}]--> Resultado: '{resultado}'")

        if not resultado:
            print("Nota: El corte devolvio una cadena vacia (revisa los indices)")
    except ValueError as e:
        print(f"Error: {e}")
```

Encuentra la posicion de una subcadena

```python
cadena = input("Ingresa una cadena: ")
if cadena.strip():
    print(f"Que subcadena buscamos dentro de: {cadena}?")
    cadena_buscar = input("Ingresa la subcadena: ")
    
    indice = cadena.find(cadena_buscar)

    if indice != -1: #En Python, cualquier número distinto de cero (incluyendo el -1) se evalúa como True
        print(f"Encontrado, {cadena_buscar} en la posicion {indice}")        
    else:
        print(f"Cadena {cadena_buscar} no encontrada", indice)
else:
    print(f"Error: ingresa una cadena correcta")
```

Cuenta cuántas veces aparece una letra o palabra

```python
error_vacio = "Error: La cadena principal esta vacia."
error_busqueda = "Error: No ingresaste nada para buscar."

cadena = input("Ingresa una cadena: ")
if cadena.strip():
    print(f"¿Qué letra a palabra quieres contar dentro de: '{cadena}'?")
    busqueda = input("Ingresa la búsqueda: ")

    if busqueda != "":        
        contador = cadena.lower().count(busqueda.lower())

        print(f"Analizando {busqueda}.....")
        print(f"Cantidad de concidencias: '{contador}'")
    else:
        print(f"{error_busqueda}")
else:
    print(f"{error_vacio}")
```

Comprueba si una cadena empieza o termina con cierto texto.

```python
error = "Error: "

cadena = input("Ingresa una cadena: ")

if cadena != "":
    buscar_principio = input("¿Con qué textro deberia empezar?: ")    
    buscar_final = input("¿Con qué texto deberia terminar?: ")
    
    resultado_inicio = f"Si está '{buscar_principio}'" if cadena.lower().startswith(buscar_principio.lower()) else f"No está '{buscar_principio}'"
    resultado_final = f"Si está '{buscar_final}'" if cadena.lower().endswith(buscar_final.lower()) else f"No está '{buscar_final}'"

    print("-"*30)

    print(f"En el principio de '{cadena}' : {resultado_inicio}")
    print(f"En el final de '{cadena}' : {resultado_final}")
    
else:
    print(f"{error}")
```

```python
alfabeticos = "Rosanela" #Solo letras (A-Z).
numericos = "3300"
alfabeto_numeros = "Rosanela,desdecomienzosdel2024" #Solo letras y números. Los espacios o comas (,) no lo son.
espacio = " "

print(f"'{alfabeticos}': {'Es puramente alfabetos' if alfabeticos.isalpha() else 'Contiene espacios o símbolos'}")
print(f"'{numericos}': {'Es puramente numérica' if numericos.isdecimal() else 'Contiene letras o espacios'}")
print(f"'{alfabeto_numeros}': {'Es solo letras y números' if alfabeto_numeros.isalnum() else 'Tiene símbolos o espacios'}")
print(f"'{espacio}': {'La cadena es solo espacios' if espacio.isspace() else 'La cadena tiene texto o está vacía'}")
```
