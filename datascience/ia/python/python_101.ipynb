1. Introducción a los tipos de datos en python
Un tipo de dato dentro de python es definido segun este sea del tipo: numerico, texto, lógico, etc.

```python
print( 5 + 10)
print("5" + "10")
```

1.2 Python es de tipado dinámico.
Python no requiere que declares el tipo de datos explicitamente de una variable, el interprete es el que infiere según el valor asignado.
Esto se conoce como tipado dinámico

```python
x = 10 # x es un entero (int)
print(type(x))

x = "Hola" # ahora x es un string (str)
print(type(x))
```

Python presenta tambien un tipado fuerte porque no permite mezclar tipos incompatibles.

```python
print(10 + "10") # ❌ Error: no se puede sumar int con str
```

1.3 ¿Por qué existen los tipos de datos?
Dentro de python los tipos de datos permiten:
- Almacenar valores correctamente.
- Realizar cálculos.
- Controlar estructuras lógicas.
- Organizar información compleja en estructuras  avanzadas.
- Determinar cómo se comportan las variable en memoria.
Python tiene varios tipos de datos importantes los principales son:

```python
# Números
# int --> Enteros
iva = 13
print(type(iva))

# float --> Decimales
sueldo = 2475.8
print(type(sueldo))

# complex --> Números complejos
distancia = 1j
print(type(distancia))
```

```python
# Textos
# str --> cadenas de caracteres
saludo = "Hola"
print(type(saludo))

# True / False --> Booleanos
es_administrador = False
print(type(es_administrador))
```

```python
# Tipos de colecciones
# list --> listas(mutable)
dias_trabajo = ["Lunes", "Martes"]
print(type(dias_trabajo))

# tuple --> Tuplas (inmutables)
procesadores= ("AMD","Intel","SnapDragon")
print(type(procesadores))

# set --> Conjuntos (sin orden, sin duplicados)
disco_duro = {"Wd","Segate"}
print(type(disco_duro))

# dict --> Diccionarios (pares clave - valor)
lenguaje = {"PHP":"Laravel", "PYTHON":"TkInter"}
print(type(lenguaje))
```

1.5 Saber el tipo de dato con type()
Python mediante esta funcion informa que tipo de dato es.
type(valor)

```python
print(type(10))         # class int
print(type(3.14))       # class float
print(type("Hola"))     # class str
print(type(True))       # class bool
print(type([1,2,3]))    # class list
```

1.6 Conversion entre tipos (CASTING)
Dentro de python puedes transformar un tipo a otro

```python
numero_cadena = "10"
casting_numero_cadena = int(numero_cadena)
print(type(numero_cadena), " -> " ,type(casting_numero_cadena))
print("")

pi_cadena = "3.14"
casting_pi_cadena = float(pi_cadena)
print(type(pi_cadena), " -> " ,type(casting_pi_cadena))
print("")

numero = 123
casting_numero = str(numero)
print(type(numero), " -> " ,type(casting_numero))
print("")

valor = 1
casting_valor = bool(valor)
print(type(valor), " -> ",type(casting_valor))
```
